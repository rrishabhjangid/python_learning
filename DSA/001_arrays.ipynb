Move Zeroes

```python
a1=[0,1,0,3,12]

insert_pos=0

for i in range(len(a1)):
    if a1[i]!=0:
        a1[insert_pos]=a1[i]
        insert_pos+=1

for i in range(insert_pos,len(a1)):
    a1[i]=0

print(a1)
```

```python
# best time to buy and sell a stock
prices = [8, 7, 2, 5, 1, 3]
min_price=float("inf")
profit=0
max_profit=0

for i in range(len(prices)):
    if prices[i]<min_price:
        min_price=prices[i]
    profit=prices[i]-min_price
    if profit>max_profit:
        max_profit=profit

print(f"Profit: {max_profit}")
```

```python
#finding maximum and minimum in a single loop
nums=[8,3,12,5,7]
min=float("inf")
max=0
for num in nums:
    if num<=min:
        min=num
    elif num>max:
        max=num
    else:
        continue

print(f"Maximum: {max}\nMinimum: {min}")
```

```python
#finding the missing number (1 to N)
def missing_number(nums):
    n=len(nums)+1

    seen={}
    for i in range(n):
        seen[i+1]=0

    for num in nums:
        seen[num]=1

    missing=next(key for key,value in seen.items() if value==0)
    return missing


#better approach using sum formula
def missing_no(nums):
    n=len(nums)+1

    expected_sum=(n*(n+1))/2
    actual_sum=sum(nums)

    return expected_sum-actual_sum

nums=[6,2,1,5,3]
print(missing_number(nums))
print(missing_no(nums))
```

```python
#rotate array by k without creating a new arrau
def rotate_array(nums,k):
    left=0
    right=len(nums)-1
    temp=None
    if k>len(nums):
        k=k%len(nums)
    while left<right:
        temp=nums[left]
        nums[left]=nums[right]
        nums[right]=temp
        left+=1
        right-=1

    left=0
    right=k-1
    temp=None
    while left<right:
        temp=nums[left]
        nums[left]=nums[right]
        nums[right]=temp
        left+=1
        right-=1

    left=k
    right=len(nums)-1
    temp=None
    while left<right:
        temp=nums[left]
        nums[left]=nums[right]
        nums[right]=temp
        left+=1
        right-=1
    return nums

nums = [1,2,3]
k=4
rotate_array(nums,k)
```

```python
#union
a=[1,4,6,3,1]
b=[4,6,1,7,9,4]

union=list(set(a)|set(b))

print(union)

union1=[]
for num in a:
    if num not in union1:
        union1.append(num)
for num in b:
    if num not in union1:
        union1.append(num)

print(union1)
```

```python
#intersection
a=[1,4,6,3,1]
b=[4,6,1,7,9,4]

intersection=list(set(a)&set(b))
print(intersection)

intersection1=[]
for num in a:
    if num in b and num not in intersection1:
        intersection1.append(num)

print(intersection1)
```

```python
#kadane's algorithm (maximum subarray sum)
nums=[-2,1,-3,4,-1,2,1,-5,4]
theMax=float('-inf')
max=0

for i in range(len(nums)):
    max+=nums[i]
    if max>theMax:
        theMax=max
    if max<0:
        max=0

print(theMax)
```
